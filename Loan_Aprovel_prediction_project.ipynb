## **Import Required Libraries**

```python
import pandas as pd
import numpy as np
```

## **Load the Dataset**

```python
df=pd.read_csv("loan_approval_1000.csv")
```

```python
df
```

## **Handle Missing Values**

```python
df.isna().sum()
```

```python
len(df)
```

```python
df["Employment_Type"].value_counts()
```

```python
mu = df["Income"].mean()
mod = df["Employment_Type"].mode()[0]
```

```python
mu
```

```python
mod
```

```python
df["Employment_Type"].mode()
```

```python
df["Income"]=df["Income"].fillna(mu)
df["Employment_Type"]=df["Employment_Type"].fillna(mod)
```

```python
df.isna().sum()
```

##**Encode Categorical Variables & Data Preprocessing**

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

df["Employment_Type"] = le.fit_transform(df["Employment_Type"])
df
```

```python
le.classes_
```

##**Split Data into Training and Testing Sets**

```python
x = df.drop("Loan_Approved",axis=1)
y = df["Loan_Approved"]

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)

```

##**Build the Machine Learning Model**
##**Train the Logistic Regression Model**

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)
```

##**Model Prediction**

```python
y_pred =lr.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

##**Model Evaluation**

```python
from sklearn.metrics import classification_report

print (classification_report(y_test,y_pred))
```

##** Hyperparameter Tuning**

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
